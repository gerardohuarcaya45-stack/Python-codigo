Ejercicio 1:
Mostrar la tabla de multiplicar del 5. Hacerlo en ambos formatos.

```python
#Modelo: tema5_ejercicio.py

t = 5
i = 0
while (True):
  i += 1 #<-- contador
  print(t, " x ", i, " = ", (t*i))
  if (i == 12):
    break
```
