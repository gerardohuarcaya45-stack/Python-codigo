```python
# Modulo: tema03_ejercicio2.py

pasaje = int(input("Ingrese codigo del pasaje: "))
precio = float(input("Ingrese precio del pasaje: "))
cantidad = int(input("Ingrese cantidad de pasajes: "))

icompra = precio * cantidad

idescuento = icompra * 0.05
if (icompra >= 400):
  idescuento = icompra * 0.15

ipagar = icompra - idescuento

caramelo = 0
if (cantidad >= 6):
  caramelo = cantidad * 3
else:
  caramelo = cantidad * 2


print("---------------------------------------")
print("Detalle de Compras:")
print("---------------------------------------")
print("Codigo del pasaje             : ", pasaje)
print("Precio               : ", precio)
print("Cantidad             : ", cantidad)
print("---------------------------------------")
print("Importe de Compra    : ", icompra)
print("Importe de Descuento : ", idescuento)
print("Importe a Pagar      : ", ipagar)
print("Cantidad de caramelos: ", caramelo)




```

```python
# Modulo: tema03_ejercicio1.py

producto = input("Ingrese Producto: ")
precio = float(input("Ingrese Precio: "))
cantidad = int(input("Ingrese Cantidad: "))

icompra = precio * cantidad

idescuento = icompra * 0.05
if (cantidad > 10):
    idescuento = icompra * 0.15

ipagar = icompra - idescuento

obsequio = "Cuaderno"
if (ipagar >= 200):
    obsequio = "Agenda"


print("---------------------------------------")
print("Detalle de Compras:")
print("---------------------------------------")
print("Producto             : ", producto)
print("Precio               : ", precio)
print("Cantidad             : ", cantidad)
print("---------------------------------------")
print("Importe de Compra    : ", icompra)
print("Importe de Descuento : ", idescuento)
print("Importe a Pagar      : ", ipagar)
print("Obsequio             : ", obsequio)

```

```python
# Modulo: tema03_ejercicio3.py

codigo = int(input("Ingrese Codigo: "))
cantidad = int(input("Ingrese Cantidad: "))

precio = 0.0
if (codigo == 101):
    precio = 17.5
elif (codigo == 102):
    precio = 25.0
elif (codigo == 103):
    precio = 15.5
else:
    print("Error en el Código")

icompra = precio * cantidad

idescuento = 0.0
if(cantidad < 11):
    idescuento = icompra * 0.05
elif (cantidad >= 11 and cantidad < 21):
    idescuento = icompra * 0.075
elif (cantidad >= 21 and cantidad < 31):
    idescuento = icompra * 0.1
else:
    idescuento = icompra * 0.125

ipagar = icompra - idescuento
obsequio = "Un Lapicero"
if (ipagar > 250):
    obsequio = "Es una Agenda"

print("---------------------------------------")
print("Detalle de compra")
print("---------------------------------------")
print("Codigo                 : ", codigo)
print("Precio                 : ", precio)
print("Cantidad               : ", cantidad)
print("---------------------------------------")
print("Importe de Compra      : ", icompra)
print("Importe de Descuento   : ", idescuento)
print("Importe a Pagar        : ", ipagar)
print("Obsequio               : ", obsequio)
```
