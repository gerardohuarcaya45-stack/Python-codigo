```python
#Modulo: listas_operaciones.py

L = [10, 25, 15, 5, 50, 2, 8]
print("L: ", L )
#Obtener el tamaño de la lista.
print("Tamaño de lista : ", len(L))
#Si en caso no existiera len
contador = 0
for i in range(7):
  contador += 1
print("Tamaño de lista : ", contador)
print()

#Obtener la sumatoria de los elementos de una lista
print("Sumatoria de elementos: ", sum(L))
#¿si en caso no existe sum?
# L = [10, 25, 15, 5, 50, 2, 8]
acum = 0
for i in range(7):
  acum += L[i]
print("Sumatoria de elementos: ", acum)

#Obtener el valor maximo de los elementos de una lista
print("El valor máximo de lista: ", max (L))
#¿Si en caso no exista max?
# L = [10, 25, 15, 5, 50, 2, 8]
mayor = 0
for i in range(7):
  if(L[i] >= mayor):
    mayor = L[i]
print("El valor máximo de lista: ", mayor)
print()
#Obtener el valor minimo de los elementos de una lista
print("El valor minimo de lista: ", min(L))
#¿Si en caso no existe min?
# L = [10, 25, 15, 5, 50, 2, 8]
menor = 99999999
for i in range(7):
  if(L[i] <= menor):
    menor = L[i]
print("El valor minimo de lista: ", menor)
print()

#Obtener el promedio aritmetico de los datos de lista
print("Promedio Aritmetico: ", sum(L)/len(L))
#¿Sino existiera sum y len?
print("Promedio Aritmético: ", acum/contador)
```
