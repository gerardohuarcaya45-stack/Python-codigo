```python
#Modulo: casualidad.py
inicio=int(input("Inicio: "))
final= int(input("Final: "))
pasos=int(input("Pasos: "))

tabla = int(input("Ingrese Numero de tabla: "))

contador = inicio
for i in range(inicio, final+1, pasos):
  print(tabla, "x", contador, "=", tabla*contador)
  contador += pasos
```

```python
#Modulo: tema4_ejercicio8.py
"""
Elabora un programa, el cual permita el ingreso de: el precio de un producto y la cantidad soliccitada, el programa debe acumular todos los importes de los productos, y visualizar los al final, de 3 productos
"""
acumulador = 0.0
for i in range(3):
  print()
  producto = input("Ingrese Producto: ")
  precio = float(input("Ingrese Precio: "))
  cantidad = int(input("Ingrese Cantidad: "))
  importe = precio * cantidad
  acumulador += importe

print()
igv = acumulador * 0.18
total = acumulador + igv
print("Total Acumulado: ", acumulador)
print("Igv (18%):", igv)
print("Total Pagar: ", total)
```
